```python
# Inicialización de la base de datos (lista vacía)
minero = []

print("--- BIENVENIDO AL SIMULADOR DE MINERO CON IA ---")
print("El robot está encendido y listo para explorar.")

# Bucle principal de ejecución
while True:
    print("\n-------------------------------------------")
    print("ESTADO ACTUAL:", inventario)
    print("Acciones: [1] Minar | [2] Procesar Material | [3] Salir")

    opcion = input("Selecciona una acción (1-3): ")

    if opcion == "1":
        print("\n> El minero está escaneando el terreno...")
        mineral = input("¿Qué mineral encontró la IA? (oro/carbon/piedra): ")

        # Lógica de decisión para agregar a la lista
        if mineral == "oro":
            print("¡Logro desbloqueado! Material valioso detectado.")
            inventario.append("oro")
        elif mineral == "carbon":
            print("Combustible recolectado.")
            inventario.append("carbon")
        elif mineral == "piedra":
            print("Material común detectado.")
            inventario.append("piedra")
        else:
            print("Error: Material desconocido. La IA no sabe cómo recolectarlo.")

    elif opcion == "2":
        print("\n> Analizando inventario para procesamiento...")

        # La IA revisa si la lista está vacía
        if inventario == []:
            print("Alerta: El inventario está vacío. No hay nada que procesar.")
        else:
            # Lógica simple de IA: Prioriza procesar el oro si existe
            if "oro" in inventario:
                print("IA Decisión: Procesando Oro para convertirlo en Lingote.")
                inventario.remove("oro")
                print("Resultado: Se ha creado un Lingote de Oro.")
            elif "carbon" in inventario:
                print("IA Decisión: Quemando carbón para obtener energía.")
                inventario.remove("carbon")
                print("Resultado: Energía del robot al 100%.")
            else:
                print("IA Decisión: Descartando piedras sin valor.")
                inventario.remove("piedra")
                print("Resultado: Espacio liberado en inventario.")

    elif opcion == "3":
        print("Apagando sistemas del minero... ¡Adiós!")
        break # Rompe el ciclo While True

    else:
        print("Comando no reconocido. Reintentando...")
```
