```python
# ==============================================
# IA DE MINERO VIRTUAL - VERSIÓN TABULAR
# Nivel: Principiante
# Uso de listas, if, elif, else y while True
# ==============================================

# Lista vacía al inicio
minero = []

print("===================================")
print("      IA DE MINERO VIRTUAL")
print("===================================")

while True:
    print("\n------ MENÚ PRINCIPAL ------")
    print("1. Registrar hallazgo")
    print("2. Ver inventario de hallazgos (Tabla)")
    print("3. Buscar consejo técnico")
    print("4. Eliminar último registro")
    print("5. Salir")

    opcion = input("Seleccione una opción: ")

    # ==========================================
    # REGISTRAR HALLAZGO
    # ==========================================
    if opcion == "1":
        nombre = input("Nombre del operario: ")
        material = input("Tipo de material encontrado: ")

        minero.append([nombre, material])

        print("\nHallazgo registrado correctamente.")

        # Respuestas inteligentes
        if material == "oro":
            print("Consejo IA: Material de alta pureza. Guardar en bóveda.")
        elif material == "carbon":
            print("Consejo IA: Útil para energía. Enviar a la caldera.")
        elif material == "hierro":
            print("Consejo IA: Material resistente. Enviar a forja.")
        elif material == "diamante":
            print("Consejo IA: ¡Extrema rareza! Usar herramienta de precisión.")
        else:
            print("Consejo IA: Material no clasificado. Analizar dureza.")

    # ==========================================
    # VER HALLAZGOS (EN FORMATO TABLA)
    # ==========================================
    elif opcion == "2":
        if minero == []:
            print("\nNo hay registros en el inventario.")
        else:
            print("\n" + "="*35)
            print(f"{'ID':<4} | {'OPERARIO':<15} | {'MATERIAL':<10}")
            print("-" * 35)

            # Recorremos la lista para imprimir cada fila de la tabla
            contador = 0
            for registro in minero:
                # registro[0] es el nombre, registro[1] es el material
                print(f"{contador:<4} | {registro[0]:<15} | {registro[1]:<10}")
                contador = contador + 1
            print("="*35)

    # ==========================================
    # BUSCAR CONSEJO TÉCNICO
    # ==========================================
    elif opcion == "3":
        consulta = input("\nEscriba el nombre del material para asesoría: ")

        if consulta == "oro":
            print("IA: Requiere lavado por batea para separar impurezas.")
        elif consulta == "carbon":
            print("IA: Cuidado con el polvo inflamable en la excavación.")
        elif consulta == "hierro":
            print("IA: Se recomienda usar explosivos de baja potencia.")
        elif consulta == "diamante":
            print("IA: Se encuentra generalmente en chimeneas de kimberlita.")
        elif consulta == "gas":
            print("IA: ¡Peligro! Evacuar túnel inmediatamente.")
        else:
            print("IA: No tengo datos sobre ese material.")

    # ==========================================
    # ELIMINAR ÚLTIMO REGISTRO
    # ==========================================
    elif opcion == "4":
        if minero == []:
            print("\nNo hay registros para eliminar.")
        else:
            minero.pop() # Elimina el último elemento de la lista
            print("\nÚltimo registro eliminado correctamente.")

    # ==========================================
    # SALIR DEL PROGRAMA
    # ==========================================
    elif opcion == "5":
        print("\nSaliendo de la IA del minero virtual. ¡Hasta pronto!")
        break # Sale del ciclo while True

    else:
        print("\nOpción no válida. Por favor, intente de nuevo.")
```
