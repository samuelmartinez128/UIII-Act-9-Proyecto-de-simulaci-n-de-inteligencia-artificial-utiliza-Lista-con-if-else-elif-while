```python
# Inicialización de la base de datos
minero = []  # Lista principal que contendrá los diccionarios

print("================================================")
print("   SISTEMA DE GESTIÓN MINERA INTELIGENTE 4.0   ")
print("================================================")

while True:
    # --- INTERFAZ DEL MENU ---
    print("\n--- MENÚ PRINCIPAL DE OPERACIONES ---")
    print("1. [REGISTRAR] Nuevo minero en la flota")
    print("2. [VISUALIZAR] Ver tabla de datos actual")
    print("3. [IA] Ejecutar análisis y optimización")
    print("4. [SALIR] Desconectar sistema")
    print("------------------------------------------------")

    opcion = input("Seleccione una opción (1-4): ")

    if opcion == "1":
        print("\n> INICIANDO PROTOCOLO DE REGISTRO...")
        # Reiniciamos el diccionario para el nuevo registro
        minero_actual = {}

        nombre = input("Ingrese el nombre del minero: ")
        mineral = input("Tipo de mineral a extraer (oro/plata/carbon): ")

        # Guardamos la información en el diccionario
        minero_actual["nombre"] = nombre
        minero_actual["mineral"] = mineral
        minero_actual["nivel"] = 1

        # Agregamos el diccionario a la lista de base de datos
        minero.append(minero_actual)
        print(f"¡CONFIRMADO! El minero {nombre} ha sido añadido.")

    elif opcion == "2":
        print("\n--- BASE DE DATOS: TABLA DE MINEROS ---")
        # Verificamos si la lista está vacía
        if minero == []:
            print("AVISO: No hay mineros registrados en la memoria.")
        else:
            # Encabezados de la tabla con formato alineado
            print(f"{'ID':<4} | {'NOMBRE':<15} | {'MINERAL':<10} | {'NIVEL':<5}")
            print("-" * 45)

            # Contador manual para el ID (ya que no usamos funciones avanzadas)
            id_contador = 1
            for m in minero:
                print(f"{id_contador:<4} | {m['nombre']:<15} | {m['mineral']:<10} | {m['nivel']:<5}")
                id_contador = id_contador + 1

    elif opcion == "3":
        print("\n--- IA: PROCESANDO DATOS EN TIEMPO REAL ---")
        if minero == []:
            print("ERROR: La lista está vacía. IA inactiva.")
        else:
            print("> Analizando rendimiento de la flota...")
            for m in minero:
                # La IA toma decisiones según el mineral que extrae el minero
                if m["mineral"] == "oro":
                    print(f"[IA] {m['nombre']} detectó material valioso. Nivel +5.")
                    m["nivel"] = m["nivel"] + 5
                elif m["mineral"] == "plata":
                    print(f"[IA] {m['nombre']} tiene rendimiento estable. Nivel +2.")
                    m["nivel"] = m["nivel"] + 2
                else:
                    print(f"[IA] {m['nombre']} extrae material común. Nivel +1.")
                    m["nivel"] = m["nivel"] + 1
            print("> Optimización completa.")

    elif opcion == "4":
        print("\n> Cerrando sesión... Todos los datos se han guardado.")
        print("¡Hasta la próxima, operador!")
        break # Detiene el While True

    else:
        print("\n[ALERTA] Opción inválida. Por favor, use números del 1 al 4.")
```
